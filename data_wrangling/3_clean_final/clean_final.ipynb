```python
import pandas as pd

# Read the dataset
df = pd.read_csv('input.csv')

# Function to remove duplicate genres from a genre string
def remove_duplicate_genres(genre_string):
    if pd.isna(genre_string):
        return genre_string
    
    # Split by comma, strip whitespace, and remove duplicates while preserving order
    genres = genre_string.split(',')
    unique_genres = []
    seen = set()
    
    for genre in genres:
        genre = genre.strip()
        if genre and genre.lower() not in seen:
            unique_genres.append(genre)
            seen.add(genre.lower())
    
    return ','.join(unique_genres)

# Apply the function to the genre column
df['genre'] = df['genre'].apply(remove_duplicate_genres)

print("Duplicate genres removed successfully!")
print(f"Dataset shape: {df.shape}")
```

```python
# Filter out games where all_styles is 0 (no playtime data)
df_with_playtime = df[df['all_styles'] > 0]
df = df_with_playtime.copy()

print(f"Original dataset shape: {df.shape}")
print(f"Dataset after removing games with no all_styles playtime: {df_with_playtime.shape}")
print(f"Removed {df.shape[0] - df_with_playtime.shape[0]} games with no playtime data")
```

```python
# Remove rows where genre is NaN or empty
df = df.dropna(subset=['genre'])
print(f"Dataset shape after removing rows with missing genres: {df.shape}")
```

```python
# Save the cleaned data
df.to_csv('output.csv', index=False)
```
